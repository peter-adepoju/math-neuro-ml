## Exercise 6 — Current versus Conductance Inhibition (10 points)

An LIF assumes **current** input. Real synapses open channels: the current $g_s(E_s - V)$ depends on the voltage. A constant inhibitory conductance $g$ (in units of $1/R_m$, dimensionless) with reversal at rest ($E_s = E_L$, "shunting") gives
$$\tau_m\frac{dV}{dt} = -(V - E_L) - g(V - E_L) + R_m I = -(1 + g)(V - E_L) + R_m I,$$
an LIF with $\tau' = \tau_m/(1 + g)$ and $R' = R_m/(1 + g)$.

Lesson 21's Exercise 7 found, for this neuron, $\tau_m \approx 34$ ms and $R_m \approx 141$ MΩ from early-window fits.

```python
# PROVIDED: load the recording and a spike detector.
import os, urllib.request
DATA_FILE7 = 'patch_171116sh_0018_steps.npz'
for cand in [DATA_FILE7, os.path.join('..', '..', 'data', DATA_FILE7)]:
    if os.path.exists(cand):
        path7 = cand; break
else:
    urllib.request.urlretrieve('https://raw.githubusercontent.com/peter-adepoju/math-neuro-ml/e8996ac071a2a16c48d5b544ec3470816eecf49f/data/' + DATA_FILE7, DATA_FILE7)
    path7 = DATA_FILE7
d7 = np.load(path7)
V7 = d7['v_cmV'] / 100.0                                                       # (17, 30000) mV
I7 = d7['i_pA'].astype(float)                                                  # (17, 30000) pA
fs7 = float(d7['fs_hz']); t7 = np.arange(V7.shape[1]) / fs7                    # (30000,) s
ep1 = (int(round(0.1469 * fs7)), int(round(0.6469 * fs7)))                     # epoch 1 sample range
ep2 = (int(round(1.6469 * fs7)), int(round(2.1469 * fs7)))                     # epoch 2 sample range
I_ep7 = I7[:, ep1[0] + 10]                                                     # (17,) pA

def detect_spikes(v, fs, level=-10.0, dvdt_thr=10.0):
    # Upward crossings of `level` mV (indices); threshold = V where dV/dt first exceeds dvdt_thr mV/ms in the 3 ms before; trough after each spike.
    up = np.where((v[1:] > level) & (v[:-1] <= level))[0] + 1
    dvdt = np.diff(v) * fs / 1000.0
    thr, trough = [], []
    for k, u in enumerate(up):
        w = np.arange(max(0, u - int(0.003 * fs)), u); c = w[dvdt[w] > dvdt_thr]
        thr.append(v[c[0]] if len(c) else np.nan)
        nxt = up[k + 1] if k + 1 < len(up) else len(v); trough.append(v[u:nxt].min())
    return up, np.array(thr), np.array(trough)
print(V7.shape, "currents:", I_ep7.astype(int).tolist())
```
