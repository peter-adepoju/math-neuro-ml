to every neuron with at least 100 spikes in the training trials of the mouse motor-cortex data ([Steinmetz et al., 2019](https://doi.org/10.1038/s41586-019-1787-x); Neuromatch Academy version; the original data are in the public domain, CC0). For every neuron, $x_t$ is the standardised wheel speed in the same 50 ms bin. The cell below prepares `x7` and `Y7` (one column per neuron) and the helper `poisson_parts`. Do not modify it.

```python
# PROVIDED -- do not modify.
import os, urllib.request

MOP_FILE = 'steinmetz_s10_hench_mop.npz'
for candidate in [MOP_FILE, os.path.join('..', '..', 'data', MOP_FILE)]:
    if os.path.exists(candidate):
        mop_path = candidate
        break
else:
    urllib.request.urlretrieve('https://raw.githubusercontent.com/peter-adepoju/math-neuro-ml/51d88220064023b5d3240c39e19b07acc7896f44/data/' + MOP_FILE, MOP_FILE)
    mop_path = MOP_FILE
mop = np.load(mop_path)
n_tr7, n_neu7, _ = mop['spikes'].shape
counts7 = mop['spikes'].reshape(n_tr7, n_neu7, 50, 5).sum(axis=3).transpose(0, 2, 1).reshape(-1, n_neu7).astype(float)  # (17100, 447) counts per 50 ms bin
speed7 = np.abs(mop['wheel'].reshape(n_tr7, 50, 5).sum(axis=2) / 0.05).reshape(-1)                                      # (17100,) wheel speed, steps/s
train7 = np.repeat(np.arange(n_tr7) < n_tr7 // 2, 50)
Ytr = counts7[train7]                                            # (8550, 447) training counts
neurons7 = np.where(Ytr.sum(axis=0) >= 100)[0]                   # neurons with at least 100 training spikes
x7 = (speed7[train7] - speed7[train7].mean()) / speed7[train7].std()   # (8550,) standardised speed, shared by all neurons
Y7 = Ytr[:, neurons7]                                            # (8550, n_neurons) counts
C7 = np.log(Y7.mean(axis=0))                                     # (n_neurons,) fixed offsets c

def poisson_parts(w, j):
    # Loss, slope and curvature of neuron j's Poisson loss at gain w.
    lam = np.exp(C7[j] + w * x7)
    y = Y7[:, j]
    return np.mean(lam - y * (C7[j] + w * x7)), np.mean((lam - y) * x7), np.mean(lam * x7**2)

print("neurons:", neurons7.size, "  inputs:", x7.shape, "  counts:", Y7.shape)
```
